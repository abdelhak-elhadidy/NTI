```python
df['Year'] = df['Year'].fillna(df['Year'].median())
df['Price'] = df['Price'].fillna(df['Price'].median())
df['Mileage'] = df['Mileage'].fillna(df['Mileage'].mean())
df['Engine_Size'] = df['Engine_Size'].fillna(df['Engine_Size'].mean())
```

```python
df['Brand'] = df['Brand'].fillna(df['Brand'].mode()[0])
df['Fuel_Type'] = df['Fuel_Type'].fillna(df['Fuel_Type'].mode()[0])
df['Transmission'] = df['Transmission'].fillna(df['Transmission'].mode()[0])
df['Previous_Owners'] = df['Previous_Owners'].fillna(df['Previous_Owners'].mode()[0])
```

```python
df.info()
```

```python
X = df.drop('Price', axis=1)
y = df[['Price']]

categorical_cols = X.select_dtypes(include=['object']).columns

X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

X.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y,
                                   test_size=0.20, random_state=100)

print("X_train = ", X_train.shape, " y_train = ", y_train.shape)
print("X_test  = ", X_test.shape, " y_test  = ", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

my_scaler = StandardScaler()
X_train_scaled = my_scaler.fit_transform(X_train)
X_test_scaled = my_scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsRegressor


knn_model = KNeighborsRegressor(n_neighbors=4)

# Train the model using the scaled training data
knn_model.fit(X_train_scaled, y_train)
```

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

knn_model = KNeighborsRegressor(n_neighbors=5)

knn_model.fit(X_train_scaled, y_train)

y_pred_knn = knn_model.predict(X_test_scaled)

# Evaluate the model
mae_knn = mean_absolute_error(y_test, y_pred_knn)
mse_knn = mean_squared_error(y_test, y_pred_knn)
rmse_knn = np.sqrt(mse_knn)
r2_knn = r2_score(y_test, y_pred_knn)

print(f"KNN Model Performance (with scaled data):\n")
print(f"Mean Absolute Error (MAE): {mae_knn:.2f}")
print(f"Mean Squared Error (MSE): {mse_knn:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse_knn:.2f}")
print(f"R-squared (R2): {r2_knn:.2f}")
```
